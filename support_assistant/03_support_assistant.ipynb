```python
from sentence_transformers import SentenceTransformer
import chromadb
import os

# Load embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Create Chroma database
client = chromadb.Client()

collection = client.create_collection("zepto_docs")

docs_path = "support_assistant/docs"

for file in sorted(os.listdir(docs_path)):
    with open(os.path.join(docs_path, file), "r") as f:
        text = f.read()

    embedding = model.encode(text).tolist()

    collection.add(
        ids=[file],
        documents=[text],
        embeddings=[embedding]
    )

print("All documents embedded successfully!")
print("Total documents:", collection.count())
```

Create the Prompt Template

```python
PROMPT_TEMPLATE = """
ROLE:
You are a Zepto customer support assistant.

CONTEXT:
Use ONLY the information provided below.

{context}

TASK:
Answer the user's question using only the above context.

NEGATIVE CONSTRAINT:
Do NOT answer using information that is not present in the provided context.
If the answer is unavailable, clearly say that the information is not available.

FORMAT:
Return a clear and concise answer.

LENGTH:
Maximum 100 words.

--------------------
Few-shot Example

Context:
Zepto offers Basic, Pass and Pass+ memberships.

Question:
What memberships are available?

Answer:
Zepto offers three memberships:
Basic, Zepto Pass and Zepto Pass+.

--------------------

Question:
{question}

Answer:
"""

print(PROMPT_TEMPLATE)
```

Import LangGraph

```python
from typing import TypedDict

from langgraph.graph import StateGraph, END
```

Create the State

```python
class SupportState(TypedDict):
    query: str
    intent: str
    answer: str
    sources: list
    confidence: float
```

Create the first node (classify_intent)

```python
KEYWORDS = [
    "delivery",
    "return",
    "refund",
    "membership",
    "tracking",
    "cancel",
    "gift card",
    "support hours"
]

def classify_intent(state):

    query = state["query"].lower()

    if any(keyword in query for keyword in KEYWORDS):
        state["intent"] = "policy_question"
    else:
        state["intent"] = "general_question"

    return state
```

```python
test = {
    "query": "What is your refund policy?"
}

print(classify_intent(test))
```

Create retrieve_and_answer node

```python
def retrieve_and_answer(state):

    query = state["query"]

    query_embedding = model.encode(query).tolist()

    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=3
    )

    top_document = results["documents"][0][0]
    source_ids = results["ids"][0]

    state["answer"] = f"Based on the retrieved context: {top_document[:200]}"

    state["sources"] = source_ids

    state["confidence"] = 1.0

    return state
```

```python
test = {
    "query": "What is the refund policy?"
}

print(retrieve_and_answer(test))
```

Create the direct_answer node

```python
def direct_answer(state):

    state["answer"] = "I can only answer questions about Zepto policies right now."

    state["sources"] = []

    state["confidence"] = 1.0

    return state
```

```python
test = {
    "query": "Who won the IPL?"
}

print(direct_answer(test))
```

Create the StateGraph

```python
from typing import TypedDict
from langgraph.graph import StateGraph, END

class SupportState(TypedDict):
    query: str
    intent: str
    answer: str
    sources: list
    confidence: float
```

Build the graph

```python
builder = StateGraph(SupportState)

builder.add_node("classify_intent", classify_intent)
builder.add_node("retrieve_and_answer", retrieve_and_answer)
builder.add_node("direct_answer", direct_answer)
```

Add Conditional Routing

```python
def route_intent(state):
    if state["intent"] == "policy_question":
        return "retrieve_and_answer"
    else:
        return "direct_answer"


builder.set_entry_point("classify_intent")

builder.add_conditional_edges(
    "classify_intent",
    route_intent,
    {
        "retrieve_and_answer": "retrieve_and_answer",
        "direct_answer": "direct_answer",
    }
)

builder.add_edge("retrieve_and_answer", END)
builder.add_edge("direct_answer", END)
```

Compile the Graph

```python
graph = builder.compile()

print(graph)
```

```python
result = graph.invoke({
    "query": "What is the refund policy?"
})

print(result)
```

```python
result = graph.invoke({
    "query": "Who won the IPL?"
})

print(result)
```

```python
%%writefile support_assistant/README.md

# Zepto Support Assistant

## Overview

This project is a Retrieval-Augmented Generation (RAG) based customer support assistant for Zepto. It answers customer policy-related questions using a local knowledge base, ChromaDB vector search, LangGraph workflow, and FastAPI.

The project works completely offline in MOCK mode without requiring any LLM API.

---

# Features

- Document-based Retrieval Augmented Generation (RAG)
- Local embeddings using Sentence Transformers
- ChromaDB vector database
- LangGraph workflow
- FastAPI REST API
- Structured JSON output
- Offline MOCK_LLM support

---

# Project Structure

support_assistant/

├── docs/

├── chroma_db/

├── main.py

├── Dockerfile

├── requirements.txt

└── README.md

---

# RAG Pipeline Architecture

## 1. Ingestion

Eight Zepto policy documents are stored inside the docs folder.

↓

## 2. Embedding

Each document is converted into vector embeddings using:

SentenceTransformer("all-MiniLM-L6-v2")

↓

## 3. Storage

The embeddings are stored inside the ChromaDB collection:

zepto_policies

↓

## 4. Retrieval

For every policy-related question:

- Query embedding is created
- Top 3 similar documents are retrieved from ChromaDB

↓

## 5. Generation

LangGraph decides the execution path.

classify_intent

↓

If Policy Question

↓

retrieve_and_answer

↓

Return answer from retrieved context

Else

↓

direct_answer

↓

Return fixed mock response

---

# LangGraph Nodes

1. classify_intent

Determines whether the question is:

- policy_question
- general_question

using keyword matching.

---

2. retrieve_and_answer

Retrieves relevant policy documents from ChromaDB and generates an answer using the retrieved context.

---

3. direct_answer

Returns:

"I can only answer questions about Zepto policies right now."

for unrelated questions.

---

# MOCK_LLM

By default:

MOCK_LLM=1

No external LLM API is used.

Only retrieval happens.

If MOCK_LLM=0, a real LLM can optionally be integrated.

---

# Example API Calls

## Policy Question

POST /ask

Request

{
    "query":"What is the refund policy?"
}

Response

{
    "answer":"Based on the retrieved context: Grocery and perishable items may be reported...",
    "sources":["doc_02"],
    "confidence":1.0
}

---

## General Question

POST /ask

Request

{
    "query":"Who won the IPL?"
}

Response

{
    "answer":"I can only answer questions about Zepto policies right now.",
    "sources":[],
    "confidence":1.0
}

---

# Run Locally

Install dependencies

pip install -r requirements.txt

Start API

uvicorn main:app --reload

---

# Docker

Build

docker build -t zepto-support .

Run

docker run -p 7860:7860 zepto-support

The API will be available at

http://localhost:7860/docs
```

```python
%%writefile support_assistant/main.py

import os
from typing import TypedDict

import chromadb
from sentence_transformers import SentenceTransformer

from fastapi import FastAPI
from pydantic import BaseModel

from langgraph.graph import StateGraph, END

# ----------------------------
# Configuration
# ----------------------------

MOCK_LLM = os.getenv("MOCK_LLM", "1") == "1"

KEYWORDS = [
    "delivery",
    "return",
    "refund",
    "membership",
    "tracking",
    "cancel",
    "gift card",
    "support hours"
]

# ----------------------------
# Load embedding model
# ----------------------------

model = SentenceTransformer("all-MiniLM-L6-v2")

client = chromadb.PersistentClient(
    path="support_assistant/chroma_db"
)

collection = client.get_or_create_collection(
    name="zepto_policies"
)

# ----------------------------
# State
# ----------------------------

class SupportState(TypedDict):
    query: str
    intent: str
    answer: str
    sources: list
    confidence: float

# ----------------------------
# Request & Response Models
# ----------------------------

class QueryRequest(BaseModel):
    query: str


class QueryResponse(BaseModel):
    answer: str
    sources: list
    confidence: float

# ----------------------------
# FastAPI
# ----------------------------

app = FastAPI(
    title="Zepto Support Assistant"
)
# ----------------------------
# Prompt Template
# ----------------------------

PROMPT_TEMPLATE = """
ROLE:
You are a Zepto customer support assistant.

CONTEXT:
Use ONLY the information provided below.

{context}

TASK:
Answer the user's question using only the above context.

NEGATIVE CONSTRAINT:
Do NOT answer using information that is not present in the provided context.
If the answer is unavailable, clearly say that the information is not available.

FORMAT:
Return a clear and concise answer.

LENGTH:
Maximum 100 words.

Example:

Context:
Zepto offers Basic, Pass and Pass+ memberships.

Question:
What memberships are available?

Answer:
Zepto offers three memberships:
Basic, Zepto Pass and Zepto Pass+.

Question:
{question}

Answer:
"""

# ----------------------------
# Node 1 - classify_intent
# ----------------------------

def classify_intent(state: SupportState):

    query = state["query"].lower()

    if any(keyword in query for keyword in KEYWORDS):
        state["intent"] = "policy_question"
    else:
        state["intent"] = "general_question"

    return state


# ----------------------------
# Node 2 - retrieve_and_answer
# ----------------------------

def retrieve_and_answer(state: SupportState):

    query = state["query"]

    query_embedding = model.encode(query).tolist()

    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=3
    )

    documents = results["documents"][0]
    ids = results["ids"][0]

    top_document = documents[0]

    if MOCK_LLM:
        answer = f"Based on the retrieved context: {top_document[:200]}"
    else:
        # Placeholder for optional real LLM
        answer = f"Based on the retrieved context: {top_document[:200]}"

    state["answer"] = answer
    state["sources"] = ids
    state["confidence"] = 1.0

    return state
    # ----------------------------
# Node 3 - direct_answer
# ----------------------------

def direct_answer(state: SupportState):

    if MOCK_LLM:
        state["answer"] = "I can only answer questions about Zepto policies right now."
    else:
        # Placeholder for optional real LLM
        state["answer"] = "I can only answer questions about Zepto policies right now."

    state["sources"] = []
    state["confidence"] = 1.0

    return state


# ----------------------------
# Routing Function
# ----------------------------

def route_intent(state: SupportState):

    if state["intent"] == "policy_question":
        return "retrieve_and_answer"

    return "direct_answer"


# ----------------------------
# Build LangGraph
# ----------------------------

builder = StateGraph(SupportState)

builder.add_node("classify_intent", classify_intent)
builder.add_node("retrieve_and_answer", retrieve_and_answer)
builder.add_node("direct_answer", direct_answer)

builder.set_entry_point("classify_intent")

builder.add_conditional_edges(
    "classify_intent",
    route_intent,
    {
        "retrieve_and_answer": "retrieve_and_answer",
        "direct_answer": "direct_answer",
    }
)

builder.add_edge("retrieve_and_answer", END)
builder.add_edge("direct_answer", END)

graph = builder.compile()
# ----------------------------
# FastAPI Endpoint
# ----------------------------

@app.post("/ask", response_model=QueryResponse)
def ask(request: QueryRequest):

    result = graph.invoke(
        {
            "query": request.query
        }
    )

    return QueryResponse(
        answer=result["answer"],
        sources=result["sources"],
        confidence=result["confidence"]
    )


# ----------------------------
# Run FastAPI
# ----------------------------

if __name__ == "__main__":

    import uvicorn

    uvicorn.run(
        "main:app",
        host="0.0.0.0",
        port=7860,
        reload=True
    )
```
